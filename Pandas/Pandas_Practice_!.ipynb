#Q.7 Task:
 Identify the workclasses that have an average hours-per-week greater than 40 AND have a total count of individuals with a marital_status of 'Never-married' less than those with 'Married-civ-spouse'.


```python
df7 = pd.read_csv("/content/drive/MyDrive/DAV_1 assignment/Datasets/adult_census.csv", encoding = "Latin1")
df7.head(2)
```

```python
df7.columns
```

```python
# df_new = df7[df7['hours.per.week'] > 40]
# df_new.head(2)
```

```python
avg_hours = df7.groupby('workclass')['hours.per.week'].mean()
avg_hours
```

```python
df7['marital.status'].value_counts()
```

```python
workclass_stats = df7.groupby('workclass').agg(
    average_hours=('hours.per.week', 'mean'),
    never_married_count=('marital.status', lambda x: (x == 'Never-married').sum()),
    married_count=('marital.status', lambda x: (x == 'Married-civ-spouse').sum())
).reset_index()

workclass_stats
```

```python
result = workclass_stats[
    (workclass_stats['average_hours'] > 40) &
    (workclass_stats['never_married_count'] < workclass_stats['married_count'])
]
result
```

Task:
 Find the Top 3 Sub-Categories (Sub-Category) that have the highest average profit per order, but restrict the analysis only to orders placed in the 'Central' region by customers in the 'Consumer' segment.


```python
df8.shape
```

```python
# df_new = df8.where((df8['Segment']=='Consumer') & (df8['Region']=='Central')).head(16)
df_new = df8[(df8['Segment']=='Consumer') & (df8['Region']=='Central')]
df_new.head(2)
```

```python
# top_3 = df8.groupby('Sub-Category')['Profit'].mean()
# top_3
```

```python
Filter = df_new.groupby('Sub-Category').agg(Total_Profit=('Profit','sum'), order_count=('Order ID','count'))
Filter.head(2)
```

```python
Filter["avg_per_order"] = Filter["Total_Profit"]/Filter["order_count"]
Filter.head()
```

```python
top3 = Filter.sort_values(by = 'avg_per_order', ascending = False)
top3[['avg_per_order']].head(3)
```
