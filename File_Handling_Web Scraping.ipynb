**4.** Read a CSV file containing student details and display all students whose marks > 75

```python
import csv
# Creating a sample CSV file
filename = "students.csv"

with open(filename, mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["id", "name", "marks"])
    writer.writerow([1, "Alice", 82])
    writer.writerow([2, "Bob", 68])
    writer.writerow([3, "Charlie", 90])
    writer.writerow([4, "David", 75])
    writer.writerow([5, "Eva", 88])

with open('students.csv', 'r') as f:
    print(f.read())
```

```python
"""
csv.reader() – Reads CSV file row by row as a list.
csv.writer() – Writes data to a CSV file row by row.

csv.DictReader() – Reads CSV data as dictionaries (column names as keys).
csv.DictWriter() – Writes CSV data using dictionaries.
"""
```

```python
with open('students.csv', 'r') as f:
    reader = csv.reader(f)
    next(reader) # Skip the header row
    for row in reader:
        if int(row[2]) > 75: ## 'marks' is at index 2
            print(row)
```

```python
import csv
with open('students.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        if int(row['marks']) > 75:
            print(row)
```

**5.** Write a program to calculate the average of a numeric column from a CSV file.

```python
with open('students.csv', 'r') as f:
    reader = csv.DictReader(f)
    total = 0
    count = 0
    for row in reader:
        total += int(row['marks'])
        count += 1
    print('Average:', total/count)
```

**6.** Read data from a CSV file and write the filtered output to a new CSV file.

```python
with open('students.csv', 'r') as f:
    reader = csv.DictReader(f)
    with open('students_filtered.csv', 'w', newline='') as f2:
        columns = ['id', 'name', 'marks']
        writer = csv.DictWriter(f2, fieldnames=columns)
        writer.writeheader()
        for row in reader:
            if int(row['marks']) > 75:
                writer.writerow(row)

with open('students_filtered.csv', 'r') as f:
    print(f.read())
```

**7.** Read a JSON file and display all key–value pairs in a readable format.

```python
"""
JSON Library Functions :-
json.load(file)
Reads JSON data from a file and converts it into a Python object.

json.loads(string)
Reads JSON data from a string.

json.dump(data, file)
Writes Python data to a JSON file.

json.dumps(data)
Converts Python data into a JSON-formatted string.

Parameters :-
indent – Formats JSON output for readability
sort_keys – Sorts keys alphabetically
"""
```

```python
import json
data = {
    "id": 101,
    "name": "Alice",
    "age": 20,
    "marks": {
        "math": 85,
        "science": 90,
        "english": 78
    },
    "passed": True
}

with open("student.json", "w") as file:
    json.dump(data, file, indent=4)
with open("student.json", "r") as file:
    print(file.read())
```

**9.** Read an XML file and extract specific elements (e.g., title, author).

```python
from bs4 import BeautifulSoup

with open('dict.xml', 'r', encoding='utf-8') as file:
    xml_content = file.read()

# Parse the XML content
soup = BeautifulSoup(xml_content, 'xml')

for book in soup.find_all('book'):
    title = book.find('title').text
    author = book.find('author').text
    print(f"Title: {title}, Author: {author}")
```

**11.** Write a program to read a log file and count the number of ERROR, WARNING, and INFO messages.

What are Logs?

Logs are records of events that happen in a computer system, application, or server. They are usually written to a file (called a log file) to help developers and system administrators monitor, debug, and analyze the behavior of software.

Key Points About Logs:

Purpose:

Track what is happening in a system.

Debug errors and issues.

Monitor performance or security events.

Log Levels:
Logs often have levels to indicate their severity:

INFO → General information about system operations. Example: "Server started at 12:00 PM".

WARNING → Something unexpected happened, but the system can continue. Example: "Disk space low".

ERROR → Something went wrong and may require attention. Example: "Failed to connect to database".

DEBUG → Detailed information for developers to debug. Example: "Variable x has value 10".

CRITICAL → Very serious error that might stop the program. Example: "System crash".

Format:
Logs are usually plain text and can include:

Timestamp

Severity level

Message

Sometimes module or function name

```python
log_file_path = "krishna.log"

error_count = 0
warning_count = 0
info_count = 0

try:
    with open(log_file_path, "r") as file:
        for line in file:
            if "ERROR" in line:
                error_count += 1
            elif "WARNING" in line:
                warning_count += 1
            elif "INFO" in line:
                info_count += 1

    print(f"ERROR messages: {error_count}")
    print(f"WARNING messages: {warning_count}")
    print(f"INFO messages: {info_count}")

except FileNotFoundError:
    print(f"Error: The file '{log_file_path}' does not exist.")
```

**12.** Read a large text file and display the top 10 most frequent bi-grams.

```python
from collections import Counter

with open('simple.txt', 'r') as f:
    text = f.read()
    words = text.split()
    bigrams = [(words[i] + ' ' + words[i+1]) for i in range(len(words)-1)]
    bigram_counts = Counter(bigrams)
    top_10_bigrams = bigram_counts.most_common(10)
    print(top_10_bigrams)
```

**18.** Scrape all table data from a webpage and save it as a CSV file.

```python
import requests
from bs4 import BeautifulSoup

url = "https://iiitk.ac.in/"
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')
links = soup.find_all('a')
#print(links)

with open('links.txt', 'w') as f:
    for link in links:
        href = link.get('href')
        if href:
            f.write(href + '\n')

with open('links.txt', 'r') as f:
    lines = f.readlines()  # reads all lines into a list
    for line in lines[:5]:
        print(line.strip())
```

**19.** Extract all image URLs from a webpage and display them.

```python
import requests
from bs4 import BeautifulSoup

url = "https://iiitk.ac.in/"
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')
images = soup.find_all('img')
#print(images)

for image in images:
    src = image.get('src')
    if src:
        print(src)
```

**20.** Scrape headlines from a news website and store them in a file.

```python
import requests
from bs4 import BeautifulSoup

url = "https://iiitk.ac.in/"
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')
headlines = soup.find_all('h2')

with open('headlines.txt', 'w') as f:
    for headline in headlines:
        f.write(headline.text)

with open('headlines.txt', 'r') as f:
    lines = f.readlines()
    print(lines)
    for line in lines[:]:
        print(line.strip())
```
