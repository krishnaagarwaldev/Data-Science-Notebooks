**21.** Use a public REST API to fetch user data and display it in a structured format.

```python
import requests
import json

url = "https://randomuser.me/api/"

response = requests.get(url)
data = response.json()

user_info = data['results'][0]
name = f"{user_info['name']['title']} {user_info['name']['first']} {user_info['name']['last']}"
email = user_info['email']
phone = user_info['phone']
country = user_info['location']['country']

print(name)
print(email)
print(phone)
print(country)
```

**23.** Use a JSON placeholder API to retrieve posts and store them in a JSON file.

```python
import requests
import json

n = int(input("Enter the post number: "))
url = f'https://jsonplaceholder.typicode.com/todos/{n}'
#url = 'https://jsonplaceholder.typicode.com/posts'
response = requests.get(url)
data = response.json()

with open('todo.json', 'w') as f:
    json.dump(data, f, indent=4)

print("Successfully fetched data")
print(json.dumps(data, indent=4))
```
